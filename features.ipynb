**Rename Files in a folder**

```python
import os

# Function to rename files
def rename_files_in_folder(folder_path, new_name):
    # List all files in the folder
    files = os.listdir(folder_path)
    
    for count, filename in enumerate(files):
        # Get file extension
        file_extension = os.path.splitext(filename)[1]
        
        # Construct new filename with sequential numbering
        new_filename = f"{new_name}{count + 1}{file_extension}"
        
        # Form full path for old and new file names
        old_file = os.path.join(folder_path, filename)
        new_file = os.path.join(folder_path, new_filename)
        
        # Rename the file
        os.rename(old_file, new_file)
        print(f"Renamed '{filename}' to '{new_filename}'")

# Specify the folder path and new file name pattern
folder_path = "D:\WORK\ASS-VOICE\data/normal"  # Replace with your folder path
new_name = "s_"  # Replace with your desired new name

# Call the function to rename files
rename_files_in_folder(folder_path, new_name)
```
